## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Load Dataset

```python
df = pd.read_csv('sales.csv')
df.head()
```

## 3. Basic Data Exploration

```python
df.info()
```

```python
df.describe()
```

## 4. Handling Missing Values

```python
df.isnull().sum()
```

## 5. Python Loops

```python
for col in df.columns:
    print(col)
```

## 6. If-Else Conditions

```python
def categorize_sales(amount):
    if amount > 500:
        return "High"
    elif amount > 200:
        return "Medium"
    else:
        return "Low"

df['Sales_Category'] = df.iloc[:, -1].apply(categorize_sales)
df.head()
```

## 7. Pandas Operations

```python
df.groupby('Sales_Category').mean(numeric_only=True)
```

## 8. NumPy Operations

```python
np.mean(df.select_dtypes(include=np.number))
```

## 9. Data Visualization

```python
df.select_dtypes(include=np.number).hist(figsize=(10,8))
plt.show()
```

```python
df["total_price"].hist()
plt.title("Distribution of Total Sales")
plt.xlabel("Total Price")
plt.ylabel("Frequency")
plt.show()
```
