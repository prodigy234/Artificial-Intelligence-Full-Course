## 🧠 Section 1 · What Is Machine Learning?

### The Big Idea (Explained Simply)

Think about how you learned to recognise a dog. Nobody gave you a list of rules like:
- "If it has 4 legs AND fur AND barks → it is a dog"

Instead, your parents pointed at many dogs and said *"dog!"*, pointed at cats and said *"cat!"*, and over time **your brain figured out the pattern by itself**.

**Machine Learning is exactly this** — we show a computer thousands of examples with the correct answers, and the computer figures out the pattern by itself. We do not write the rules manually.

---

### Traditional Programming vs Machine Learning

```
TRADITIONAL PROGRAMMING:
  Rules (written by human) + Data  ──→  Output
  Example: IF price > 1000 THEN label = "expensive"

MACHINE LEARNING:
  Data + Correct Answers  ──→  Rules (learned automatically by computer)
  Example: show 10,000 photos labelled "cat"/"dog" → computer learns the rules itself
```

---

### The Three Types of Machine Learning

| Type | How It Learns | Real Example |
|------|--------------|-------------|
| **Supervised Learning** 👨‍🏫 | Learns from examples WITH correct answers | Spam detection, disease diagnosis |
| **Unsupervised Learning** 🔍 | Finds hidden patterns in data with NO labels | Customer grouping, anomaly detection |
| **Reinforcement Learning** 🎮 | Agent learns by trial-and-error with rewards | Chess AI, self-driving cars |

**This entire notebook covers Supervised Learning** — it is the most common type used in business and science.

---

### Key Vocabulary (Learn These!)

| Word | Simple Meaning | Example |
|------|---------------|---------|
| **Features (X)** | The inputs — what we know about each example | Petal length, blood pressure |
| **Target (y)** | What we want to predict | Species name, "malignant"/"benign" |
| **Model** | The learned pattern / mathematical formula | The rules the computer discovered |
| **Training** | Teaching the model using examples with answers | Showing the model 10,000 labelled photos |
| **Prediction** | Using the trained model on new data | "This new flower is a setosa" |
| **Accuracy** | How often the model is correct | 95% = correct 95 times out of 100 |

## 🌸 Section 3 · Our First Real Dataset — Iris Flowers

### The Story Behind This Dataset

In **1936**, British statistician Ronald Fisher collected measurements from 150 iris flowers in the Gaspé Peninsula of Canada. He measured 4 physical properties of each flower and recorded which of 3 species it was.

This dataset is still used today to teach machine learning because:
- It is **small enough** to understand completely
- It is **real data** (not made up)
- The **patterns are clear** and meaningful

### Our Goal
Can we train a computer to **automatically identify the species** of an iris flower just from its measurements? Yes — and we will do it in this section!

### The Four Measurements (Features)
```
Sepal: The green outer leaf-like part protecting the bud
Petal: The coloured inner flower petals

  ┌──────────────────────────────────┐
  │  sepal length │  sepal width     │  ← measurements of the outer part
  │  petal length │  petal width     │  ← measurements of the inner petals
  └──────────────────────────────────┘
```

### The Three Species (Target Classes)
- 🌸 **Setosa** — small, compact flower
- 🌷 **Versicolor** — medium-sized flower
- 🌺 **Virginica** — larger flower

### This Is a Classification Problem
We are predicting a **category** (which species), not a number.
Classification = predicting which group/category something belongs to.

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 2 — LOADING THE IRIS DATASET
# ═══════════════════════════════════════════════════════════════════

# Import the function that loads the Iris dataset
# sklearn comes with several datasets built-in — no internet needed!
from sklearn.datasets import load_iris

# load_iris() downloads the dataset and returns a "Bunch" object
# A Bunch is like a dictionary — it has named fields we can access
iris = load_iris()

# ─────────────────────────────────────────────────────────────────
# Let's explore what's inside this Bunch object
# ─────────────────────────────────────────────────────────────────

print("═══ IRIS DATASET — FIRST LOOK ═══")
print()

# type() tells us what kind of object something is
print(f"Object type        : {type(iris)}")
print()

# iris.data = the measurements (a 2D array: rows=flowers, columns=measurements)
# .shape returns (number_of_rows, number_of_columns)
print(f"Feature matrix (X) : {iris.data.shape}")
print(f"  → {iris.data.shape[0]} flowers measured")
print(f"  → {iris.data.shape[1]} measurements per flower")
print()

# iris.target = the species labels (a 1D array: one label per flower)
print(f"Target vector (y)  : {iris.target.shape}")
print(f"  → one species label per flower")
print()

# iris.feature_names = the names of our 4 columns
print(f"Feature names (columns):")
for i, name in enumerate(iris.feature_names):
    print(f"  Column {i}: {name}")
print()

# iris.target_names = what 0, 1, 2 actually mean
print(f"Target names (species):")
for i, name in enumerate(iris.target_names):
    print(f"  Label {i} = {name}")
print()

# Show the actual first 5 rows of data
print("First 5 rows of raw measurement data:")
print(iris.data[:5])       # [:5] means rows 0, 1, 2, 3, 4
print()
print("Their species labels:", iris.target[:5])
print("  → All 0s = all setosa in the first 5 rows")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 3 — CONVERT TO PANDAS DATAFRAME (Make It Readable)
# ═══════════════════════════════════════════════════════════════════

# Raw NumPy arrays are hard to read — no column names, no species names
# We convert to a pandas DataFrame which looks like a proper table

# pd.DataFrame() creates a table from the array
# data=    : the raw numbers
# columns= : the column header names
iris_df = pd.DataFrame(
    data    = iris.data,
    columns = iris.feature_names
)

# Add a readable species name column
# iris.target gives [0, 0, 0, ..., 1, 1, ..., 2, 2, 2]
# We map: 0 → 'setosa', 1 → 'versicolor', 2 → 'virginica'
iris_df['species'] = [iris.target_names[label] for label in iris.target]

# Also keep the numeric label (we need this for modelling later)
iris_df['species_id'] = iris.target

# ─────────────────────────────────────────────────────────────────
# Explore the DataFrame
# ─────────────────────────────────────────────────────────────────

print("═══ IRIS AS A TABLE ═══")
print()
print(f"Shape: {iris_df.shape} → {iris_df.shape[0]} rows (flowers), {iris_df.shape[1]} columns")
print()

# .head(n) shows the first n rows — default is 5
print("First 10 rows:")
print(iris_df.head(10))
print()

# .dtypes shows the data type of each column
print("Column data types:")
print(iris_df.dtypes)
print()

# Check for missing values — VERY important in real work
# .isnull() returns True/False for each cell
# .sum() counts the Trues per column
missing = iris_df.isnull().sum()
print("Missing values per column:")
print(missing)
print()
print("Great! Zero missing values in this dataset.")
print("In real-world data, you almost always have missing values.")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 4 — EXPLORATORY DATA ANALYSIS (EDA) — STATISTICS
# ═══════════════════════════════════════════════════════════════════

# EDA = Exploratory Data Analysis
# We explore the data BEFORE building any model
# This is like reading the instruction manual before assembling furniture
# Skipping EDA is a common beginner mistake — don't do it!

print("═══ STATISTICAL SUMMARY ═══")
print()

# .describe() gives you a full statistical summary for every numeric column:
# count = how many values (should all be 150)
# mean  = average value
# std   = standard deviation (how spread out the values are)
# min   = smallest value
# 25%   = 25th percentile (25% of values are below this)
# 50%   = median (middle value)
# 75%   = 75th percentile
# max   = largest value

print(iris_df.describe().round(3))
print()

# ─────────────────────────────────────────────────────────────────
# Class distribution — how many of each species?
# ─────────────────────────────────────────────────────────────────

print("═══ CLASS DISTRIBUTION (How many of each species?) ═══")
print()

# .value_counts() counts how many times each value appears
species_counts = iris_df['species'].value_counts()
print(species_counts)
print()

# Calculate percentages
total = len(iris_df)
print("Percentages:")
for species, count in species_counts.items():
    pct = count / total * 100
    bar = "█" * int(pct / 2)
    print(f"  {species:12s}: {count} ({pct:.1f}%)  {bar}")
print()
print("This is a BALANCED dataset — equal samples per class.")
print("Balanced datasets are ideal for training — the model won't be biased.")
print()

# Group by species and compute mean for each feature
print("═══ MEAN MEASUREMENTS PER SPECIES ═══")
print()
print(iris_df.groupby('species')[iris.feature_names].mean().round(3))
print()
print("KEY INSIGHT: Petal measurements differ a LOT between species.")
print("Sepal width is less different — less useful for prediction.")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 5 — EDA VISUALIZATION: PAIRPLOT
#  (Every Feature Plotted Against Every Other Feature)
# ═══════════════════════════════════════════════════════════════════

# A pairplot is a grid of scatter plots
# Each cell in the grid shows TWO features plotted against each other
# The DIAGONAL shows the distribution of ONE feature (kernel density plot)
# Colour = species — so we can see if species separate in feature space

# If species of different colours cluster separately = good features for prediction
# If all colours mix together = those features are not useful

print("Creating pairplot... (this may take a few seconds)")
print()

# sns.pairplot() creates the entire grid automatically
pair_fig = sns.pairplot(
    iris_df,                    # our DataFrame

    hue='species',              # colour each dot by species

    # which columns to include in the plot
    vars=['sepal length (cm)', 'sepal width (cm)',
          'petal length (cm)', 'petal width (cm)'],

    diag_kind='kde',            # diagonal: kernel density estimate (smooth histogram)

    # styling for the scatter dots
    plot_kws={'alpha': 0.65,    # alpha = transparency (0=invisible, 1=solid)
              's': 35},         # s = dot size

    # custom colours for each species
    palette={'setosa':     '#E74C3C',    # red
             'versicolor': '#3498DB',    # blue
             'virginica':  '#2ECC71'}    # green
)

# Add a title to the whole figure
pair_fig.fig.suptitle(
    'Iris Dataset — Pairplot: Every Feature vs Every Feature '
    '(Colour = Species · Look for separated clusters!)',
    y=1.03, fontsize=13, fontweight='bold'
)

plt.tight_layout()
plt.savefig('tmp/iris_pairplot.png', dpi=100, bbox_inches='tight')
plt.show()

print("HOW TO READ THIS PAIRPLOT:")
print()
print("  Each row/column pair is one scatter plot.")
print("  Red = setosa, Blue = versicolor, Green = virginica")
print()
print("  WHAT TO LOOK FOR:")
print("  • Petal length vs Petal width: PERFECT separation!")
print("    Setosa is completely isolated in the bottom-left corner.")
print("  • Sepal measurements: More mixing/overlap between species.")
print("  • This tells us PETAL features will be the most useful for our model.")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 6 — EDA VISUALIZATION: BOX PLOTS
#  (Distribution of Each Feature, Split by Species)
# ═══════════════════════════════════════════════════════════════════

# A box plot shows the DISTRIBUTION of values for each group:
#
#     ─────── ← maximum (excluding outliers)
#        │
#   ┌────┤   ← 75th percentile (75% of data is BELOW this)
#   │    │
#   ├────┤   ← MEDIAN (50th percentile — the middle value)
#   │    │
#   └────┤   ← 25th percentile (25% of data is BELOW this)
#        │
#     ─────── ← minimum (excluding outliers)
#       ⚬    ← outlier (a value unusually far from the rest)
#
# BIG SEPARATION between species boxes = USEFUL feature for prediction
# SMALL SEPARATION (overlapping boxes) = less useful feature

# Create a figure with 2 rows and 2 columns of subplots
# 4 features = 4 box plots
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

# Add a title to the whole figure
fig.suptitle(
    'Iris Dataset — Box Plots: How Each Feature Varies by Species '
    '(Wider separation between boxes = better predictor!)',
    fontsize=13, fontweight='bold'
)

# Define colors for each species (consistent with pairplot)
species_colors = ['#E74C3C', '#3498DB', '#2ECC71']  # red, blue, green

# Plot each feature in its own subplot
features_to_plot = ['sepal length (cm)', 'sepal width (cm)',
                    'petal length (cm)', 'petal width (cm)']

for ax, feature in zip(axes.flatten(), features_to_plot):
    # Collect data for each species
    data_by_species = [
        iris_df[iris_df['species'] == sp][feature].values
        for sp in iris.target_names
    ]

    # ax.boxplot() draws the box plot
    # patch_artist=True fills the boxes with colour
    bp = ax.boxplot(
        data_by_species,
        labels=iris.target_names,
        patch_artist=True,       # fill boxes with colour
        notch=False,             # no notch in the box
        medianprops={'color': 'black', 'linewidth': 2}  # black median line
    )

    # Colour each box
    for patch, color in zip(bp['boxes'], species_colors):
        patch.set_facecolor(color)
        patch.set_alpha(0.7)

    # Labels and styling
    ax.set_title(f'Feature: {feature}', fontsize=11, fontweight='bold')
    ax.set_ylabel('Measurement (cm)', fontsize=9)
    ax.grid(axis='y', alpha=0.3, linestyle='--')

plt.tight_layout()
plt.savefig('tmp/iris_boxplots.png', dpi=100, bbox_inches='tight')
plt.show()

print("INTERPRETATION:")
print()
print("  ✅ Petal Length: HUGE separation — setosa ~1.5cm, virginica ~5.5cm")
print("  ✅ Petal Width:  HUGE separation — setosa ~0.3cm, virginica ~2.0cm")
print("  ⚠️  Sepal Length: Some separation but overlap between versicolor/virginica")
print("  ❌ Sepal Width:  Least useful — lots of overlap, setosa is actually wider!")
print()
print("CONCLUSION: Petal measurements will help the model most.")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 7 — CORRELATION HEATMAP
#  (Which Features Are Related to Each Other?)
# ═══════════════════════════════════════════════════════════════════

# Correlation measures how closely two features move together:
#  +1.0 = perfect positive correlation (when A goes up, B always goes up)
#   0.0 = no correlation (A and B are completely independent)
#  -1.0 = perfect negative correlation (when A goes up, B always goes down)
#
# WHY THIS MATTERS FOR ML:
# If two features are perfectly correlated, they contain DUPLICATE information.
# Using both doesn't help the model — it might actually hurt.

# Compute the correlation matrix (only numeric columns)
# .select_dtypes('number') picks only numeric columns
numeric_cols = iris_df.select_dtypes('number').drop('species_id', axis=1)
corr_matrix = numeric_cols.corr()

print("Correlation Matrix:")
print(corr_matrix.round(3))
print()
print("HOW TO READ: Values close to ±1 = strongly related")
print()

# Plot the heatmap
fig, ax = plt.subplots(figsize=(8, 6))

# sns.heatmap() draws a colour-coded grid
heatmap = sns.heatmap(
    corr_matrix,
    annot=True,           # show numbers in each cell
    fmt='.2f',            # format numbers to 2 decimal places
    cmap='RdYlGn',        # Red (negative) → Yellow (neutral) → Green (positive)
    center=0,             # centre the colour scale at 0
    square=True,          # make each cell square-shaped
    linewidths=0.5,       # thin lines between cells
    ax=ax,
    vmin=-1, vmax=1       # colour range from -1 to +1
)

ax.set_title(
    'Iris Dataset — Feature Correlation Heatmap '
    '(Green = strong positive, Red = strong negative, Yellow = no relationship)',
    fontsize=12, fontweight='bold'
)

plt.tight_layout()
plt.savefig('tmp/iris_heatmap.png', dpi=100, bbox_inches='tight')
plt.show()

print("KEY FINDINGS:")
print("  🟢 petal length & petal width: correlation = 0.96 (almost perfectly correlated!)")
print("     → They measure similar things. We only need one for most models.")
print("  🟢 petal length & sepal length: correlation = 0.87 (strongly related)")
print("  🟡 sepal width & others: near 0 (independent — adds unique information)")
````

## 🔑 Section 4 · The sklearn API — The Universal Pattern

### The Most Important Thing to Learn

Every single sklearn model — no matter how simple or complex — follows the **exact same 3-step pattern**:

```python
# ┌─────────────────────────────────────────────────────────────┐
# │  STEP 1: CREATE the model (choose algorithm + settings)     │
# │                                                             │
# │  model = AlgorithmName(setting1=value, setting2=value)      │
# │                                                             │
# ├─────────────────────────────────────────────────────────────┤
# │  STEP 2: TRAIN the model (let it learn from data)           │
# │                                                             │
# │  model.fit(X_train, y_train)                                │
# │                                                             │
# ├─────────────────────────────────────────────────────────────┤
# │  STEP 3: PREDICT (use the trained model on new data)        │
# │                                                             │
# │  predictions = model.predict(X_test)                        │
# └─────────────────────────────────────────────────────────────┘
```

**This pattern is called the Estimator API.** Once you understand it for one model, you understand it for ALL models in sklearn.

### The Train / Test Split — The Golden Rule

> ⚠️ **NEVER evaluate your model on the same data it was trained on!**

Imagine a student who reads the exam paper the night before, memorises all the answers, then scores 100% on the exam. Did they actually LEARN anything? No — they just memorised.

This is called **overfitting** — when a model memorises training data instead of learning real patterns.

**The Solution:** Split your data into two parts:
- **Training set (80%):** The model learns from this
- **Test set (20%):** We evaluate the model on data it has NEVER seen

```
Full Dataset (150 flowers)
    ↓
┌─────────────────────┬──────────┐
│   Training Set      │ Test Set │
│   (120 flowers)     │ (30 flowers)│
│   Model learns here │ We test here│
└─────────────────────┴──────────┘
```

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 8 — TRAIN/TEST SPLIT
#  Splitting our data safely before any modelling
# ═══════════════════════════════════════════════════════════════════

# Import the function that handles the split
from sklearn.model_selection import train_test_split

# ─────────────────────────────────────────────────────────────────
# Prepare X (features) and y (target)
# ─────────────────────────────────────────────────────────────────

# X = feature matrix — all measurement columns
# We take ONLY the measurement data, NOT the species columns
X = iris.data    # shape: (150, 4) — 150 flowers, 4 measurements each

# y = target vector — the species we want to predict
y = iris.target  # shape: (150,)  — one label per flower (0, 1, or 2)

print(f"X shape (features) : {X.shape}")
print(f"y shape (labels)   : {y.shape}")
print()

# ─────────────────────────────────────────────────────────────────
# Split the data
# ─────────────────────────────────────────────────────────────────

# train_test_split() randomly divides the data
#
# test_size=0.2  → 20% goes to test set, 80% to training
# random_state=42 → seed for the random number generator
#                   Same seed = same split every time you run the code
#                   This makes your results REPRODUCIBLE
# stratify=y  → IMPORTANT! Ensures each species appears proportionally
#               in BOTH train and test sets
#               Without this, you might accidentally get all setosa in training
#               and none in testing — which would be unfair

X_train, X_test, y_train, y_test = train_test_split(
    X,              # feature matrix
    y,              # labels
    test_size=0.2,  # 20% for testing
    random_state=42,# reproducible random split
    stratify=y      # preserve class balance in both sets
)

# ─────────────────────────────────────────────────────────────────
# Verify the split
# ─────────────────────────────────────────────────────────────────

print("═══ TRAIN/TEST SPLIT RESULTS ═══")
print()
print(f"Original dataset    : {len(X)} samples (100%)")
print(f"Training set        : {len(X_train)} samples  (80%)")
print(f"Test set            : {len(X_test)} samples  (20%)")
print()

# Verify that class balance is preserved
print("Class distribution — Training Set:")
for species_id, name in enumerate(iris.target_names):
    count = np.sum(y_train == species_id)
    pct = count / len(y_train) * 100
    print(f"  {name:12s}: {count} samples ({pct:.1f}%)")

print()
print("Class distribution — Test Set:")
for species_id, name in enumerate(iris.target_names):
    count = np.sum(y_test == species_id)
    pct = count / len(y_test) * 100
    print(f"  {name:12s}: {count} samples ({pct:.1f}%)")

print()
print("✅ Both sets have balanced classes — stratify=y worked!")
````

## 🔄 Section 5 · Data Preprocessing — Preparing Data for ML

### Why Preprocessing Matters

Before feeding data to a machine learning model, we often need to **transform it** into a better format. This is called preprocessing.

### Feature Scaling — The Most Common Step

Imagine you have two features:
- **Income** ranging from ₦50,000 to ₦5,000,000
- **Age** ranging from 20 to 65

Without scaling, the income feature has numbers thousands of times larger than age. Many algorithms use **distance between points** (like KNN) or **gradients** (like logistic regression). If income numbers are 10,000× larger than age numbers, the income feature will dominate completely — even if age is equally important!

**StandardScaler solves this** by transforming each feature to have:
- Mean (average) = 0
- Standard deviation = 1

Formula: `scaled_value = (original_value - mean) / standard_deviation`

### ⚠️ THE MOST IMPORTANT RULE IN ALL OF MACHINE LEARNING

> **ALWAYS fit the scaler on training data ONLY.**
> **Then use that same scaler to transform both training AND test data.**

**Why?** If you fit the scaler on all data (including test), the model has indirectly "peeked" at the test set before evaluation. Your results will be falsely optimistic. This mistake is called **data leakage**.

```
WRONG (data leakage!):          CORRECT:
scaler.fit(X_all)               scaler.fit(X_train)        ← train data only
X_scaled = transform(X_all)     X_train_s = transform(X_train)  ← use training stats
                                X_test_s  = transform(X_test)   ← same stats on test
```

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 9 — FEATURE SCALING WITH STANDARDSCALER
# ═══════════════════════════════════════════════════════════════════

from sklearn.preprocessing import StandardScaler

# ─────────────────────────────────────────────────────────────────
# STEP 1: Create the scaler object
# ─────────────────────────────────────────────────────────────────
scaler = StandardScaler()
# At this point, the scaler knows nothing yet
# It's like a blank ruler waiting to be calibrated

# ─────────────────────────────────────────────────────────────────
# STEP 2: FIT the scaler on TRAINING DATA ONLY
# .fit() computes the mean and standard deviation for each column
# It ONLY looks at X_train — NEVER X_test!
# ─────────────────────────────────────────────────────────────────
scaler.fit(X_train)

print("═══ WHAT THE SCALER LEARNED FROM TRAINING DATA ═══")
print()
print("Mean (average) of each feature in the training set:")
for name, mean_val in zip(iris.feature_names, scaler.mean_):
    print(f"  {name:25s}: {mean_val:.4f} cm")
print()
print("Standard deviation of each feature in the training set:")
for name, std_val in zip(iris.feature_names, scaler.scale_):
    print(f"  {name:25s}: {std_val:.4f} cm")

# ─────────────────────────────────────────────────────────────────
# STEP 3: TRANSFORM both train and test using the TRAINING statistics
# .transform() applies: (value - mean) / std_deviation
# ─────────────────────────────────────────────────────────────────
X_train_scaled = scaler.transform(X_train)
X_test_scaled  = scaler.transform(X_test)   # ← uses training mean/std, NOT test mean/std

print()
print("═══ BEFORE vs AFTER SCALING ═══")
print()
print("Before scaling — first training flower:")
print(f"  {X_train[0]}")
print()
print("After scaling — same flower:")
print(f"  {X_train_scaled[0].round(4)}")
print()
print("After scaling, values are centred around 0, typically between -3 and +3")
print()

# Verify: scaled training data should have mean ≈ 0 and std ≈ 1
print("Verification (scaled training data):")
print(f"  Mean of each feature: {X_train_scaled.mean(axis=0).round(6)}")
print(f"  Std  of each feature: {X_train_scaled.std(axis=0).round(6)}")
print()
print("✅ Mean ≈ 0 and Std ≈ 1 for all features — scaling worked!")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 10 — THE PIPELINE: Safe Preprocessing + Modelling
# ═══════════════════════════════════════════════════════════════════

# PROBLEM with manual scaling:
# What if you forget to scale the test data? Or accidentally fit on test data?
# These are easy mistakes that silently break your model.
#
# SOLUTION: sklearn Pipeline
# A Pipeline chains steps together in the CORRECT ORDER and enforces the rules.
# During .fit():     each step is fit AND transformed (on training data)
# During .predict(): each step ONLY transforms (never re-fits, never leaks)
#
# The Pipeline makes data leakage IMPOSSIBLE by design.

from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier  # our first model (explained in Section 6)

# Build a pipeline with two steps
# Each step is a tuple: ('name_you_choose', estimator_object)
# The name can be anything — just make it descriptive
pipe_demo = Pipeline([
    ('scaler', StandardScaler()),                    # Step 1: scale features
    ('model',  KNeighborsClassifier(n_neighbors=5)) # Step 2: classify
])

print("Pipeline created!")
print()
print("Pipeline steps:")
for step_name, step_object in pipe_demo.steps:
    print(f"  '{step_name}': {step_object.__class__.__name__}")
print()

# Train the ENTIRE pipeline in ONE line
# Internally: scaler.fit_transform(X_train) → knn.fit(result, y_train)
pipe_demo.fit(X_train, y_train)

# Predict in ONE line (using ORIGINAL unscaled X_test!)
# Internally: scaler.transform(X_test) → knn.predict(result)
# The pipeline handles scaling automatically
y_pred_demo = pipe_demo.predict(X_test)

# Quick accuracy check
from sklearn.metrics import accuracy_score
demo_acc = accuracy_score(y_test, y_pred_demo)

print(f"Pipeline accuracy: {demo_acc:.4f} ({demo_acc*100:.1f}%)")
print()
print("BENEFITS of using a Pipeline:")
print("  ✅ Data leakage is IMPOSSIBLE by design")
print("  ✅ Everything in one object — easy to save, share, deploy")
print("  ✅ Works directly with cross-validation and hyperparameter tuning")
print("  ✅ Less code, fewer bugs")
print()
print("We will use Pipelines throughout this notebook.")
````

## 🔍 Section 6 · K-Nearest Neighbours (KNN) — Your First ML Algorithm

### The Intuition (In Real Life)

Imagine you just moved to a new city in Nigeria and you want to know what the local neighbourhood is like. You don't ask everyone in the city — you ask your **5 nearest neighbours**. If 4 of them say "this is a safe area" and 1 says "it's dangerous", you conclude it is probably safe. You vote by **majority** among the nearest people.

**KNN works exactly this way:**
1. For a new flower we want to classify, find its **K nearest training flowers** (using distance)
2. Look at their species labels
3. **Majority vote** — the most common label wins

### The Algorithm Step-by-Step

```
New flower measurements → [5.1, 3.5, 1.4, 0.2] cm

Training data:
  Flower 1: [4.9, 3.0, 1.4, 0.2] → setosa    ─── distance = 0.22 (nearest!)
  Flower 2: [7.0, 3.2, 4.7, 1.4] → versicolor ── distance = 4.1
  Flower 3: [5.0, 3.4, 1.5, 0.2] → setosa    ─── distance = 0.14 (nearest!)
  Flower 4: [6.3, 2.5, 5.0, 1.9] → virginica ── distance = 5.2
  Flower 5: [4.6, 3.1, 1.5, 0.2] → setosa    ─── distance = 0.55 (nearest!)
  ...

For K=3: Top 3 nearest = setosa, setosa, setosa
Prediction: SETOSA (3 votes vs 0)
```

### The Key Hyperparameter: K

| K value | Problem |
|---------|---------|
| K = 1 | Overfitting — one noisy point can change the prediction |
| K = 100 | Underfitting — ignores local patterns, too many voters |
| K = 5 to 15 | Usually a good starting range |

### Distance Formula (Euclidean Distance)
For two points with 4 features:
```
distance = √[(x₁-x₁')² + (x₂-x₂')² + (x₃-x₃')² + (x₄-x₄')²]
```
This is just the Pythagorean theorem extended to 4 dimensions!

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 11 — K-NEAREST NEIGHBOURS: COMPLETE WALKTHROUGH
# ═══════════════════════════════════════════════════════════════════

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay

# ─────────────────────────────────────────────────────────────────
# STEP 1: CREATE the model
# ─────────────────────────────────────────────────────────────────
# n_neighbors = K (how many neighbours to consider for voting)
# metric = how to measure distance between points
#   'euclidean' = straight-line distance (most common)
#   'manhattan' = sum of absolute differences (like city blocks)
knn = KNeighborsClassifier(
    n_neighbors = 5,          # K = 5 neighbours
    metric      = 'euclidean' # distance formula
)

print("Model created:", knn)
print()

# ─────────────────────────────────────────────────────────────────
# STEP 2: TRAIN (fit) the model
# ─────────────────────────────────────────────────────────────────
# For KNN, 'training' just means memorising the training data
# There is no complex calculation during fit()
# The real work happens during predict()

knn.fit(X_train_scaled, y_train)
# X_train_scaled = our scaled training features
# y_train        = the correct species labels

print("✅ Model trained on", len(X_train_scaled), "flowers")
print()

# ─────────────────────────────────────────────────────────────────
# STEP 3: PREDICT on the test set
# ─────────────────────────────────────────────────────────────────
y_pred_knn = knn.predict(X_test_scaled)

print("Predictions for first 15 test flowers:")
print()
print(f"  {'Flower':<8} {'Predicted':>15} {'Actual':>15} {'Correct?':>10}")
print(f"  {'─'*8} {'─'*15} {'─'*15} {'─'*10}")

for i in range(15):
    predicted = iris.target_names[y_pred_knn[i]]
    actual    = iris.target_names[y_test[i]]
    correct   = "✅" if predicted == actual else "❌"
    print(f"  {i+1:<8} {predicted:>15} {actual:>15} {correct:>10}")
print()

# ─────────────────────────────────────────────────────────────────
# STEP 4: EVALUATE
# ─────────────────────────────────────────────────────────────────
knn_acc = accuracy_score(y_test, y_pred_knn)
print(f"═══ KNN ACCURACY ═══")
print()
print(f"  Correct predictions : {int(knn_acc * len(y_test))} out of {len(y_test)}")
print(f"  Accuracy            : {knn_acc:.4f}  ({knn_acc*100:.1f}%)")
print()
print(f"  In other words: for {knn_acc*100:.1f}% of flowers the model had never seen,")
print(f"  it correctly identified the species just from measurements.")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 12 — UNDERSTANDING EVALUATION METRICS
#  (More than just accuracy!)
# ═══════════════════════════════════════════════════════════════════

# ACCURACY alone is not always enough!
# Example: Imagine 95% of patients are healthy, 5% have cancer.
# A model that ALWAYS says "healthy" would have 95% accuracy — but it
# would miss EVERY cancer case! That is dangerous.
#
# We need metrics that tell us MORE about where errors happen.

print("═══ CLASSIFICATION REPORT ═══")
print()

# classification_report() gives detailed metrics for EACH class:
# Precision: Of all flowers I called 'setosa', what fraction really were setosa?
# Recall:    Of all real setosa flowers, what fraction did I catch?
# F1-Score:  Harmonic mean of precision and recall (balance between the two)
# Support:   How many actual samples of each class in the test set

print(classification_report(
    y_test,                          # actual correct labels
    y_pred_knn,                      # model's predictions
    target_names=iris.target_names   # replace 0,1,2 with species names
))

print()
print("HOW TO READ THE CLASSIFICATION REPORT:")
print()
print("  PRECISION = TP / (TP + FP)")
print("    'When I say setosa, how often am I right?'")
print("    A precision of 1.00 means: every time I predicted setosa, it really was setosa")
print()
print("  RECALL = TP / (TP + FN)")
print("    'Of all actual setosa flowers, how many did I find?'")
print("    A recall of 1.00 means: I found every single setosa in the test set")
print()
print("  F1-SCORE = 2 × (Precision × Recall) / (Precision + Recall)")
print("    A balanced metric that considers both precision and recall")
print("    Use this when you want one number that captures both")
print()
print("  SUPPORT = how many actual flowers of that species were in the test set")
print("    We have 10 of each (because stratify=y gave us equal splits)")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 13 — CONFUSION MATRIX
#  The Most Visual Way to See Where Your Model Makes Mistakes
# ═══════════════════════════════════════════════════════════════════

# A confusion matrix shows:
#   ROWS    = the ACTUAL correct species
#   COLUMNS = what the model PREDICTED
#   DIAGONAL = CORRECT predictions (we want these numbers HIGH)
#   OFF-DIAGONAL = WRONG predictions (we want these ZERO)
#
# Example for 3 classes:
#
#              Predicted:
#              setosa  versicolor  virginica
# Actual:
# setosa    [   10,       0,          0   ]  ← all 10 setosa correctly identified
# versicolor[    0,       9,          1   ]  ← 1 versicolor was wrong (called virginica)
# virginica [ 0,          0,         10   ]  ← all 10 virginica correctly identified

cm_knn = confusion_matrix(y_test, y_pred_knn)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('KNN — Confusion Matrix and Prediction Results', fontsize=13, fontweight='bold')

# ── Left: Confusion Matrix ──────────────────────────────────────
disp = ConfusionMatrixDisplay(
    confusion_matrix = cm_knn,
    display_labels   = iris.target_names
)
disp.plot(
    ax       = axes[0],
    cmap     = 'Blues',      # colour scheme: white → blue (more correct = darker)
    colorbar = False
)
axes[0].set_title('Confusion Matrix (Diagonal = correct, Off-diagonal = errors)',
                  fontsize=11, fontweight='bold')

# ── Right: Actual vs Predicted Bar Chart ────────────────────────
x = np.arange(len(iris.target_names))   # positions on x-axis
width = 0.35                             # width of each bar

# Count actual and predicted per class
actual_counts    = [np.sum(y_test == i) for i in range(3)]
predicted_counts = [np.sum(y_pred_knn == i) for i in range(3)]

bars1 = axes[1].bar(x - width/2, actual_counts,    width, label='Actual',    color='#3498DB', alpha=0.8)
bars2 = axes[1].bar(x + width/2, predicted_counts, width, label='Predicted', color='#E74C3C', alpha=0.8)

axes[1].set_xticks(x)
axes[1].set_xticklabels(iris.target_names)
axes[1].set_ylabel('Count'); axes[1].set_title('Actual vs Predicted Counts', fontsize=11, fontweight='bold')
axes[1].legend(); axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('tmp/knn_evaluation.png', dpi=100, bbox_inches='tight')
plt.show()

print("READING THE CONFUSION MATRIX:")
print()
for i, species in enumerate(iris.target_names):
    correct = cm_knn[i, i]
    total   = cm_knn[i].sum()
    errors  = total - correct
    if errors == 0:
        print(f"  {species:12s}: {correct}/{total} correct — PERFECT!")
    else:
        wrong_classes = [(iris.target_names[j], cm_knn[i,j]) for j in range(3) if j!=i and cm_knn[i,j]>0]
        wrong_str = ', '.join([f"{n}({c})" for n, c in wrong_classes])
        print(f"  {species:12s}: {correct}/{total} correct — {errors} mistake(s) → confused with {wrong_str}")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 14 — FINDING THE BEST K VALUE
#  Understanding the Bias-Variance Tradeoff
# ═══════════════════════════════════════════════════════════════════

# BIAS: Error from being too simple / ignoring patterns (UNDERFITTING)
# VARIANCE: Error from being too sensitive to training data (OVERFITTING)
# GOAL: Find the K value that gives the best BALANCE on NEW data

# Let's test K = 1 through K = 25
k_range          = range(1, 26)
train_accuracies = []    # how well model does on training data (data it saw)
test_accuracies  = []    # how well model does on test data (data it never saw)

for k in k_range:
    # Create and train KNN with this value of K
    knn_test = KNeighborsClassifier(n_neighbors=k)
    knn_test.fit(X_train_scaled, y_train)

    # Training accuracy (score on data the model was trained on)
    train_acc = accuracy_score(y_train, knn_test.predict(X_train_scaled))
    train_accuracies.append(train_acc)

    # Test accuracy (score on NEW data the model has never seen)
    test_acc = accuracy_score(y_test, knn_test.predict(X_test_scaled))
    test_accuracies.append(test_acc)

# Find best K for test data
best_k_idx = test_accuracies.index(max(test_accuracies))
best_k = list(k_range)[best_k_idx]
best_test_acc = max(test_accuracies)

# ── Plot ────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(11, 5))

ax.plot(list(k_range), train_accuracies, 'o-',
        color='#3498DB', linewidth=2, markersize=6, label='Training Accuracy')
ax.plot(list(k_range), test_accuracies, 's-',
        color='#E74C3C', linewidth=2, markersize=6, label='Test Accuracy')

# Mark the best K
ax.axvline(x=best_k, color='green', linestyle='--', alpha=0.8, linewidth=2)
ax.annotate(f'Best K = {best_k} Test Acc = {best_test_acc:.2f}',
            xy=(best_k, best_test_acc),
            xytext=(best_k + 2, best_test_acc - 0.03),
            fontsize=10, color='green',
            arrowprops={'arrowstyle': '->', 'color': 'green'})

# Shade overfitting and underfitting regions
ax.axvspan(1, 3, alpha=0.05, color='red', label='Overfitting zone')
ax.axvspan(20, 25, alpha=0.05, color='orange', label='Underfitting zone')

ax.set_xlabel('K (Number of Neighbours)', fontsize=12)
ax.set_ylabel('Accuracy', fontsize=12)
ax.set_title('KNN: Effect of K on Training vs Test Accuracy '
             '(Finding the sweet spot between overfitting and underfitting)',
             fontsize=12, fontweight='bold')
ax.legend(fontsize=10); ax.grid(True, alpha=0.3)
ax.set_xlim([0, 26]); ax.set_ylim([0.85, 1.02])
ax.set_xticks(list(k_range))

plt.tight_layout()
plt.savefig('tmp/knn_k_search.png', dpi=100, bbox_inches='tight')
plt.show()

print("WHAT THIS GRAPH TEACHES:")
print()
print(f"  K = 1:   Training accuracy = 100%! (memorises every point perfectly)")
print(f"           Test accuracy is LOWER — it memorised noise too! OVERFITTING")
print()
print(f"  K = 2-4: Still some overfitting")
print()
print(f"  K = {best_k}:  Best test accuracy = {best_test_acc:.2%} ← SWEET SPOT")
print()
print(f"  K = 20+: Both accuracies drop — model too generalised. UNDERFITTING")
print()
print("The gap between blue (train) and red (test) lines shows overfitting.")
print("We want the lines CLOSE TOGETHER and the test line as HIGH as possible.")
````

## 📊 Section 7 · Cross-Validation — Getting a Trustworthy Score

### The Problem with a Single Train/Test Split

When we split 150 flowers into 120 training and 30 test, we are relying on just 30 flowers to judge the model. What if those 30 flowers happened to be the "easy" ones? Or the "hard" ones? Our score could be misleading.

### The Solution: K-Fold Cross-Validation

Instead of one split, we make **K splits** and evaluate K times, rotating which part is the test set each time. Every sample gets tested exactly once.

```
K = 5 folds:

Fold 1: [  TEST  ][TRAIN][TRAIN][TRAIN][TRAIN] → Score 1 = 97%
Fold 2: [TRAIN][  TEST  ][TRAIN][TRAIN][TRAIN] → Score 2 = 95%
Fold 3: [TRAIN][TRAIN][  TEST  ][TRAIN][TRAIN] → Score 3 = 98%
Fold 4: [TRAIN][TRAIN][TRAIN][  TEST  ][TRAIN] → Score 4 = 96%
Fold 5: [TRAIN][TRAIN][TRAIN][TRAIN][  TEST  ] → Score 5 = 97%

Final Result: Mean = 96.6%  ±  Std = 1.0%
              └── 'reliable score'    └── 'how stable?'
```

**The mean gives a more reliable estimate.** The standard deviation tells you how consistent the model is across different data splits.

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 15 — 5-FOLD CROSS-VALIDATION
# ═══════════════════════════════════════════════════════════════════

from sklearn.model_selection import cross_val_score, StratifiedKFold

# StratifiedKFold ensures class balance in EVERY fold
# n_splits=5  → 5 folds
# shuffle=True → randomise order before splitting
# random_state=42 → reproducible randomisation
cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# For cross-validation we MUST use a Pipeline
# The Pipeline ensures the scaler is refit fresh in each fold
# If we manually pre-scaled, the scaler would be fit on ALL data — data leakage!
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

pipe_cv = Pipeline([
    ('scaler', StandardScaler()),
    ('knn',    KNeighborsClassifier(n_neighbors=5))
])

# cross_val_score() runs the entire pipeline through all 5 folds
# It returns an array of 5 scores — one per fold
# scoring='accuracy' → we measure accuracy in each fold

scores = cross_val_score(
    pipe_cv,        # the pipeline (model to evaluate)
    X,              # ALL features (cross-val handles the splitting!)
    y,              # ALL labels
    cv=cv_strategy, # how to split (5-fold stratified)
    scoring='accuracy'
)

print("═══ 5-FOLD CROSS-VALIDATION RESULTS ═══")
print()
print("Individual fold scores:")
for fold_num, score in enumerate(scores, 1):
    bar = "█" * int(score * 30)
    print(f"  Fold {fold_num}: {score:.4f}  {bar}")
print()
print(f"  Mean accuracy : {scores.mean():.4f}  ({scores.mean()*100:.1f}%)")
print(f"  Std deviation : {scores.std():.4f}  (how much scores vary across folds)")
print(f"  Min score     : {scores.min():.4f}")
print(f"  Max score     : {scores.max():.4f}")
print()
print("INTERPRETATION:")
print(f"  Our model achieves {scores.mean()*100:.1f}% ± {scores.std()*100:.1f}% accuracy")
print(f"  The low standard deviation ({scores.std():.4f}) means the model is CONSISTENT")
print(f"  — it doesn't rely on getting lucky with which data is in the test set.")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 16 — COMPARING MULTIPLE MODELS WITH CROSS-VALIDATION
# ═══════════════════════════════════════════════════════════════════

# Now we compare FIVE different algorithms on the same data
# This is called a MODEL COMPARISON or BENCHMARK

from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier

# Dictionary of models to compare
# Each value is a pipeline with scaler + classifier
models_to_compare = {
    'K-Nearest Neighbours (K=5)' : Pipeline([('s', StandardScaler()), ('m', KNeighborsClassifier(n_neighbors=5))]),
    'Decision Tree (depth=4)'    : Pipeline([('s', StandardScaler()), ('m', DecisionTreeClassifier(max_depth=4, random_state=42))]),
    'Logistic Regression'        : Pipeline([('s', StandardScaler()), ('m', LogisticRegression(max_iter=500, random_state=42))]),
    'Support Vector Machine'     : Pipeline([('s', StandardScaler()), ('m', SVC(kernel='rbf', C=1.0, random_state=42))]),
    'Random Forest (100 trees)'  : Pipeline([('s', StandardScaler()), ('m', RandomForestClassifier(n_estimators=100, random_state=42))]),
}

# Store results for plotting
comparison_results = {}

print(f"{'Model':<30} {'Mean Acc':>10} {'Std Dev':>9}  {'Fold Scores'}")
print("─" * 85)

for model_name, pipeline in models_to_compare.items():
    # Run 5-fold cross-validation for this model
    cv_scores = cross_val_score(pipeline, X, y, cv=cv_strategy, scoring='accuracy')

    comparison_results[model_name] = cv_scores

    # Format the fold scores for display
    fold_str = '  '.join([f"{s:.3f}" for s in cv_scores])
    print(f"{model_name:<30} {cv_scores.mean():>10.4f} {cv_scores.std():>9.4f}  [{fold_str}]")

print()
print("NOTES:")
print("  Mean = reliability estimate (higher = better)")
print("  Std  = consistency (lower = more stable across different data splits)")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 17 — VISUALISE MODEL COMPARISON
# ═══════════════════════════════════════════════════════════════════

fig, ax = plt.subplots(figsize=(12, 6))

model_names  = list(comparison_results.keys())
colors_models = ['#3498DB', '#E74C3C', '#2ECC71', '#9B59B6', '#F39C12']

for i, (model_name, scores) in enumerate(comparison_results.items()):
    # Plot each fold's score as a dot
    ax.scatter(
        [i] * 5,      # x-position (same for all 5 folds of this model)
        scores,        # y-position (the 5 fold scores)
        color=colors_models[i], s=80, zorder=5, alpha=0.8
    )
    # Plot the mean as a horizontal line
    ax.hlines(scores.mean(), i - 0.35, i + 0.35,
              color=colors_models[i], linewidth=3, zorder=6)
    # Shaded band for ± 1 standard deviation
    ax.fill_between([i - 0.35, i + 0.35],
                    scores.mean() - scores.std(),
                    scores.mean() + scores.std(),
                    alpha=0.15, color=colors_models[i])

ax.set_xticks(range(len(model_names)))
ax.set_xticklabels(model_names, rotation=20, ha='right', fontsize=10)
ax.set_ylabel('Accuracy', fontsize=12)
ax.set_ylim([0.88, 1.02])
ax.set_title('5-Fold Cross-Validation Results for All Models '
             '(Dots = individual fold scores · Line = mean · Shading = ±1 std dev)',
             fontsize=12, fontweight='bold')
ax.grid(axis='y', alpha=0.3, linestyle='--')

# Add value labels for means
for i, (_, scores) in enumerate(comparison_results.items()):
    ax.text(i, scores.mean() + 0.005, f'{scores.mean():.3f}',
            ha='center', va='bottom', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.savefig('tmp/model_comparison.png', dpi=100, bbox_inches='tight')
plt.show()

# Find the best model
best_model_name = max(comparison_results, key=lambda k: comparison_results[k].mean())
best_score = comparison_results[best_model_name].mean()
print(f"Best model: {best_model_name}")
print(f"Best score: {best_score:.4f} ({best_score*100:.1f}%)")
````

## 🌳 Section 8 · Decision Trees — The Model You Can Actually Read

### Why Decision Trees Are Special

Every other ML algorithm is a "black box" — you put in data, you get a prediction, but you cannot easily explain WHY. 

A Decision Tree is different: **you can read every single decision it makes**, like following a flowchart.

### How a Decision Tree Works

The algorithm repeatedly asks: **"Which feature, split at which threshold, best separates the classes?"**

```
Is petal length ≤ 2.45 cm?
├── YES → 🌸 SETOSA (100% certain)
└── NO  → Is petal width ≤ 1.75 cm?
          ├── YES → Is petal length ≤ 4.95 cm?
          │         ├── YES → 🌷 VERSICOLOR
          │         └── NO  → 🌺 VIRGINICA (with some errors)
          └── NO  → 🌺 VIRGINICA (high confidence)
```

### The Gini Impurity
At each node, the algorithm measures **Gini Impurity** — how mixed the classes are:
- **Gini = 0.0** = Pure node (only one class) — PERFECT
- **Gini = 0.5** = Maximum mixed (50/50 split between 2 classes) — WORST

The algorithm finds the split that **reduces Gini impurity the most**.

### Overfitting Problem
If we let the tree grow without limits, it will eventually create one leaf per training sample — 100% training accuracy but terrible on new data. This is why we use `max_depth` to limit growth.

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 18 — DECISION TREE: TRAIN AND INSPECT
# ═══════════════════════════════════════════════════════════════════

from sklearn.tree import DecisionTreeClassifier, export_text, plot_tree
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

# Scale features (good practice even for trees, though less critical)
scaler_dt = StandardScaler()
X_train_dt = scaler_dt.fit_transform(X_train)
X_test_dt  = scaler_dt.transform(X_test)

# Create and train a shallow tree (max_depth=3 keeps it readable)
dt = DecisionTreeClassifier(
    max_depth   = 3,       # limit depth to prevent overfitting (and keep it readable)
    random_state= 42       # reproducible splits
)
dt.fit(X_train_dt, y_train)

print("═══ DECISION TREE STATISTICS ═══")
print()
print(f"  Maximum depth set : 3 levels")
print(f"  Actual depth used : {dt.get_depth()} levels")
print(f"  Number of leaves  : {dt.get_n_leaves()} (final prediction nodes)")
print()
print(f"  Training accuracy : {accuracy_score(y_train, dt.predict(X_train_dt)):.4f}")
print(f"  Test accuracy     : {accuracy_score(y_test, dt.predict(X_test_dt)):.4f}")
print()

# ── TEXT REPRESENTATION OF THE TREE ────────────────────────────
print("═══ DECISION RULES LEARNED BY THE TREE ═══")
print()
print("(Read: |--- means 'branch', indentation = depth level)")
print()
print(export_text(
    dt,
    feature_names=iris.feature_names
))
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 19 — DECISION TREE: VISUAL DIAGRAM
# ═══════════════════════════════════════════════════════════════════

# plot_tree() draws the tree as a coloured diagram
# Each box (node) shows:
#   Line 1: The splitting question (e.g., petal length ≤ 0.65)
#   Line 2: Gini impurity of this node (0.0 = pure, 0.5 = mixed)
#   Line 3: samples = how many training flowers reach this node
#   Line 4: value = count per class [setosa, versicolor, virginica]
#   Line 5: class = the majority class prediction here

fig, ax = plt.subplots(figsize=(18, 9))

plot_tree(
    dt,
    feature_names = iris.feature_names,    # column names
    class_names   = iris.target_names,     # species names (not 0,1,2)
    filled        = True,                  # colour nodes by predicted class
    rounded       = True,                  # rounded box corners
    fontsize      = 9,
    ax            = ax,
    impurity      = True,                  # show gini values
    proportion    = False                  # show counts, not proportions
)

ax.set_title(
    'Iris Decision Tree (max_depth=3) '
    'How to use: Start at root → answer True/False → go Left/Right → reach prediction',
    fontsize=13, fontweight='bold'
)

plt.tight_layout()
plt.savefig('tmp/decision_tree.png', dpi=120, bbox_inches='tight')
plt.show()

print("HOW TO USE THIS TREE TO CLASSIFY A NEW FLOWER:")
print()
print("  1. Start at the TOP box (root node)")
print("  2. Answer the question: is the condition True or False?")
print("  3. Go LEFT if True, go RIGHT if False")
print("  4. Keep answering until you reach a LEAF (no more questions)")
print("  5. The leaf's 'class' is the prediction")
print()
print("EXAMPLE: New flower with petal length = 1.5 cm, petal width = 0.3 cm")
print("  Root: petal length ≤ ~0.65? → 1.5 > 0.65 → NO → go RIGHT")
print("  Next: petal width ≤ ~0.75?  → 0.3 < 0.75  → YES → go LEFT")
print("  Leaf: class = versicolor")
print()
print("The colour shows the majority class: ORANGE=setosa, GREEN=versicolor, PURPLE=virginica")
print("Darker colour = purer node (more certain prediction)")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 20 — DECISION TREE: DEPTH vs ACCURACY
#  Demonstrating Overfitting and Underfitting Visually
# ═══════════════════════════════════════════════════════════════════

depths        = range(1, 12)    # test tree depths from 1 to 11
dt_train_accs = []
dt_test_accs  = []

for depth in depths:
    # Create and train a tree with this depth
    dt_d = DecisionTreeClassifier(max_depth=depth, random_state=42)
    dt_d.fit(X_train_dt, y_train)

    # Record both training AND test accuracy
    dt_train_accs.append(accuracy_score(y_train, dt_d.predict(X_train_dt)))
    dt_test_accs.append(accuracy_score(y_test,   dt_d.predict(X_test_dt)))

fig, ax = plt.subplots(figsize=(11, 6))

# Plot both curves
ax.plot(list(depths), dt_train_accs, 'o-', color='#3498DB', linewidth=2.5,
        markersize=8, label='Training Accuracy', zorder=5)
ax.plot(list(depths), dt_test_accs, 's-', color='#E74C3C', linewidth=2.5,
        markersize=8, label='Test Accuracy', zorder=5)

# Find best test accuracy depth
best_depth = list(depths)[dt_test_accs.index(max(dt_test_accs))]
best_test  = max(dt_test_accs)
ax.axvline(best_depth, color='green', linestyle='--', linewidth=2, alpha=0.7)
ax.annotate(f'Best depth = {best_depth}',
            xy=(best_depth, best_test),
            xytext=(best_depth + 1, best_test - 0.02),
            fontsize=11, color='darkgreen',
            arrowprops={'arrowstyle':'->', 'color':'green'})

# Shade zones
ax.axvspan(1, 2, alpha=0.06, color='orange', label='Underfitting zone')
ax.axvspan(6, 11, alpha=0.06, color='red', label='Overfitting zone')

ax.set_xlabel('Max Tree Depth', fontsize=12)
ax.set_ylabel('Accuracy', fontsize=12)
ax.set_title('Decision Tree: Bias-Variance Tradeoff '
             'Training accuracy always rises with depth — test accuracy has a peak',
             fontsize=12, fontweight='bold')
ax.legend(fontsize=10); ax.grid(True, alpha=0.3)
ax.set_xticks(list(depths)); ax.set_ylim([0.80, 1.02])

plt.tight_layout()
plt.savefig('tmp/dt_depth_tradeoff.png', dpi=100, bbox_inches='tight')
plt.show()

print("THE BIAS-VARIANCE TRADEOFF (Most Important Concept in ML!):")
print()
print("  Depth 1-2: Model too SIMPLE. Cannot learn all patterns.")
print("             Both train and test accuracy are low. → UNDERFITTING / HIGH BIAS")
print()
print(f"  Depth {best_depth}:   SWEET SPOT. Best balance for test data.")
print(f"             Test accuracy peaks here: {best_test:.2%}")
print()
print("  Depth 6+: Model memorises EVERY training example (train=100%)")
print("            But fails on new data (test drops). → OVERFITTING / HIGH VARIANCE")
print()
print("  KEY INSIGHT: A model that scores 100% on training data is NOT a good model!")
print("  We always care about GENERALISATION — performance on new, unseen data.")
````

## 📈 Section 9 · Logistic Regression — Finding the Best Dividing Line

### The Name is Misleading!

Despite having "regression" in its name, **Logistic Regression is a classification algorithm**. It predicts probabilities and then assigns a class based on those probabilities.

### The Core Idea

Linear Regression fits a straight line through data. What if we apply a mathematical function that squashes that line's output into a range between 0 and 1? That function is called the **sigmoid function**:

```
σ(x) = 1 / (1 + e^(-x))

Output is ALWAYS between 0 and 1
  → Can be interpreted as PROBABILITY
  → Output > 0.5 → Class 1
  → Output < 0.5 → Class 0
```

The sigmoid function looks like an S-curve:
```
1.0 ─────────────────────────────────⎤
                                  ╱  │
0.5 ─────────────────────────  ─╱   │
                              ╱      │
0.0 ─────────────────────╱───────────┘
         ────────────────→  Input value
```

### For Multi-Class Problems (3+ classes)

For Iris (3 species), sklearn automatically uses the **One-vs-Rest** or **Softmax** strategy:
- Train 3 binary classifiers: "setosa vs rest", "versicolor vs rest", "virginica vs rest"
- The class with the highest probability wins

### The Regularisation Parameter C

- **Large C** = less regularisation = model fits training data more tightly (risk of overfitting)
- **Small C** = more regularisation = model is more constrained (risk of underfitting)
- **C = 1.0** = default, usually a good starting point

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 21 — LOGISTIC REGRESSION: TRAINING AND PROBABILITIES
# ═══════════════════════════════════════════════════════════════════

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Build pipeline: scale → logistic regression
pipe_lr = Pipeline([
    ('scaler', StandardScaler()),
    ('lr',     LogisticRegression(
                   max_iter    = 1000,     # max iterations for solver to converge
                   random_state= 42,
                   C           = 1.0       # regularisation (higher = less regularised)
               ))
])

# Train
pipe_lr.fit(X_train, y_train)

print("═══ LOGISTIC REGRESSION RESULTS ═══")
print()
print(f"Training accuracy : {accuracy_score(y_train, pipe_lr.predict(X_train)):.4f}")
print(f"Test accuracy     : {accuracy_score(y_test,  pipe_lr.predict(X_test)):.4f}")
print()

# ── PROBABILITIES ───────────────────────────────────────────────
# predict_proba() gives probability for EACH class, not just the winner
# Returns a 2D array: shape (n_samples, n_classes)
# Row sums = 1.0 (probabilities must add up to 100%)

y_proba = pipe_lr.predict_proba(X_test)

print("Predicted PROBABILITIES for first 8 test flowers:")
print()
print(f"  {'#':<4} {'P(setosa)':>12} {'P(versicolor)':>15} {'P(virginica)':>13} {'Predicted':>12} {'Actual':>12} {'OK?':>4}")
print(f"  {'─'*4} {'─'*12} {'─'*15} {'─'*13} {'─'*12} {'─'*12} {'─'*4}")

y_pred_lr = pipe_lr.predict(X_test)

for i in range(8):
    probs     = y_proba[i]
    predicted = iris.target_names[y_pred_lr[i]]
    actual    = iris.target_names[y_test[i]]
    ok        = "✅" if predicted == actual else "❌"
    print(f"  {i+1:<4} {probs[0]:>12.4f} {probs[1]:>15.4f} {probs[2]:>13.4f} "
          f"{predicted:>12} {actual:>12} {ok:>4}")

print()
print("Note: Each row's probabilities sum to 1.0")
print("The model predicts the CLASS with the HIGHEST probability")
print("High confidence prediction: one probability near 1.0, others near 0.0")
print("Uncertain prediction: probabilities spread more evenly (e.g., 0.4, 0.35, 0.25)")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 22 — LOGISTIC REGRESSION: COEFFICIENTS
#  (Understanding WHAT the model learned)
# ═══════════════════════════════════════════════════════════════════

# The trained logistic regression model has COEFFICIENTS
# These are the weights (importance scores) for each feature
# POSITIVE coefficient: higher feature value → higher probability of this class
# NEGATIVE coefficient: higher feature value → lower probability of this class

# Get the classifier from the pipeline
lr_model = pipe_lr.named_steps['lr']  # access by step name

# Coefficients: shape (n_classes=3, n_features=4)
coef = lr_model.coef_

print("═══ LOGISTIC REGRESSION COEFFICIENTS ═══")
print()
print("Each row = one class. Each column = one feature.")
print("Positive = more of this feature → more likely this class")
print("Negative = more of this feature → less likely this class")
print()

coef_df = pd.DataFrame(
    coef,
    index   = iris.target_names,
    columns = iris.feature_names
)
print(coef_df.round(4))

print()
print("Example reading:")
for feature, coef_val in zip(iris.feature_names, coef[0]):
    direction = "↑ MORE likely" if coef_val > 0 else "↓ LESS likely"
    print(f"  Setosa: larger {feature:25s}: {direction} (coef={coef_val:.3f})")

print()
print("The LARGE MAGNITUDE coefficients (far from 0) tell us")
print("which features are most important for this model's decisions.")

# Visualise coefficients
fig, ax = plt.subplots(figsize=(10, 5))

x = np.arange(len(iris.feature_names))
width = 0.25
colors_coef = ['#E74C3C', '#3498DB', '#2ECC71']

for i, (species, color) in enumerate(zip(iris.target_names, colors_coef)):
    ax.bar(x + i*width, coef[i], width,
           label=species, color=color, alpha=0.8)

ax.set_xticks(x + width)
ax.set_xticklabels(iris.feature_names, rotation=15, ha='right')
ax.axhline(0, color='black', linewidth=0.8)
ax.set_ylabel('Coefficient Value', fontsize=11)
ax.set_title('Logistic Regression Coefficients '
             '(What the model thinks is important for each species)',
             fontsize=12, fontweight='bold')
ax.legend(title='Species', fontsize=10)
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('tmp/lr_coefficients.png', dpi=100, bbox_inches='tight')
plt.show()
````

## ⚔️ Section 10 · Support Vector Machine (SVM) — Finding the Perfect Boundary

### The Core Idea: Maximum Margin

Imagine you are drawing a line on a map to separate two countries. You could draw it right next to the border of one country, or right next to the other, or right in the MIDDLE with equal space on both sides.

**SVM always draws the boundary in the middle — as far as possible from both classes.** This middle ground is called the **maximum margin** boundary.

```
Data points:
  ● ● ●  |    ■ ■ ■      ← Bad boundary (too close to ●)
  ● ● ●  | ■ ■ ■         ← Also bad (too close to ■)
  ● ● ●   |   ■ ■ ■      ← Maximum margin! Equal distance from both
              ↑
         Decision boundary (maximum margin)
```

The data points closest to the boundary are called **Support Vectors** — that's where the algorithm gets its name.

### The Kernel Trick — Handling Non-Linear Data

What if the data cannot be separated by a straight line?

```
IMPOSSIBLE with a straight line:    POSSIBLE after transformation!
  ● ● ●●●                              ●     ●
  ■ ■● ■■■                            ──────────
  ■ ■■ ●■■                              ■   ■ ■
```

The **kernel trick** mathematically transforms data into a higher dimension where it IS linearly separable, without explicitly computing the expensive transformation.

| Kernel | When to Use |
|--------|------------|
| `linear` | Data is roughly linearly separable |
| `rbf` | Most common — handles curves and complex boundaries |
| `poly` | Polynomial boundaries |

### Key Parameters
- **C**: Regularisation (like in Logistic Regression — higher = fits training more tightly)
- **gamma**: Radius of influence of a training point (only for rbf/poly). `'scale'` = automatic

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 23 — SUPPORT VECTOR MACHINE: TRAINING AND EVALUATION
# ═══════════════════════════════════════════════════════════════════

from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# Build SVM pipeline
pipe_svm = Pipeline([
    ('scaler', StandardScaler()),  # SVM is VERY sensitive to scale — always scale!
    ('svm',    SVC(
                   kernel      = 'rbf',    # Radial Basis Function — most versatile
                   C           = 1.0,      # regularisation
                   gamma       = 'scale',  # automatic gamma based on data
                   random_state= 42,
                   probability = True      # enable predict_proba() — needs this flag
               ))
])

# Train
pipe_svm.fit(X_train, y_train)
y_pred_svm = pipe_svm.predict(X_test)

print("═══ SVM RESULTS ═══")
print()
print(f"Training accuracy : {accuracy_score(y_train, pipe_svm.predict(X_train)):.4f}")
print(f"Test accuracy     : {accuracy_score(y_test,  y_pred_svm):.4f}")
print()
print("Classification Report:")
print()
print(classification_report(y_test, y_pred_svm, target_names=iris.target_names))
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 24 — COMPARING KERNEL TYPES ON IRIS
# ═══════════════════════════════════════════════════════════════════

# Let's see how different kernels perform on this dataset
kernels = ['linear', 'rbf', 'poly', 'sigmoid']
kernel_results = {}

print("═══ SVM KERNEL COMPARISON ═══")
print()
print(f"{'Kernel':<12} {'Train Acc':>12} {'Test Acc':>12} {'Verdict'}")
print("─" * 60)

for kernel in kernels:
    pipe_k = Pipeline([
        ('s', StandardScaler()),
        ('svm', SVC(kernel=kernel, C=1.0, gamma='scale', random_state=42))
    ])
    pipe_k.fit(X_train, y_train)
    train_acc = accuracy_score(y_train, pipe_k.predict(X_train))
    test_acc  = accuracy_score(y_test,  pipe_k.predict(X_test))
    kernel_results[kernel] = (train_acc, test_acc)

    gap = train_acc - test_acc
    verdict = "⚠️ Some overfit" if gap > 0.03 else "✅ Good generalisation"
    print(f"{kernel:<12} {train_acc:>12.4f} {test_acc:>12.4f} {verdict}")

print()
print("The 'rbf' (Radial Basis Function) kernel is usually the best default choice.")
print("It can learn curved boundaries and handles most real-world problems well.")
````

## 🌲🌲🌲 Section 11 · Random Forest — The Power of Many Trees

### The Wisdom of Crowds

Imagine you need to estimate how many beans are in a jar. You ask one person — they say 340. You ask 100 people and average their guesses — you get 347 (the real answer is 350). **The average of many guesses is almost always better than one guess.**

This principle is called the **Wisdom of Crowds**, and Random Forest applies it to machine learning.

### How Random Forest Works

```
Build 100 Decision Trees, each with TWO sources of randomness:

  Randomness 1 — Bootstrap Sampling:
    Each tree trains on a RANDOM SAMPLE of the data (with replacement)
    Some rows appear twice, some not at all → each tree sees different data
    This is called "bagging" (Bootstrap AGGregation)

  Randomness 2 — Random Feature Selection:
    At each split, only consider a RANDOM SUBSET of features
    (not all features — just √n_features of them)
    This forces trees to find different patterns

  Final Prediction:
    Classification: Each tree votes → majority wins
    Regression:     Average all tree predictions
```

### Why This Works Better Than One Tree

Each individual tree is slightly wrong in different ways (due to randomness). When you average 100 trees, **their individual errors cancel out**, and you are left with a much more accurate and stable prediction.

### Feature Importance
Random Forest gives us a bonus: it can rank which features are most useful for prediction. This is called **feature importance**.

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 25 — RANDOM FOREST: TRAINING AND FEATURE IMPORTANCE
# ═══════════════════════════════════════════════════════════════════

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Build Random Forest pipeline
pipe_rf = Pipeline([
    ('scaler', StandardScaler()),
    ('rf',     RandomForestClassifier(
                   n_estimators = 200,    # how many trees to build
                   max_depth    = None,   # no depth limit (each tree can grow fully)
                   random_state = 42,     # reproducible
                   n_jobs       = -1      # use all available CPU cores (faster)
               ))
])

pipe_rf.fit(X_train, y_train)
y_pred_rf = pipe_rf.predict(X_test)

print("═══ RANDOM FOREST RESULTS ═══")
print()
print(f"  n_estimators (trees)  : 200")
print(f"  Training accuracy     : {accuracy_score(y_train, pipe_rf.predict(X_train)):.4f}")
print(f"  Test accuracy         : {accuracy_score(y_test, y_pred_rf):.4f}")
print()

# ── FEATURE IMPORTANCE ─────────────────────────────────────────
# .feature_importances_ gives a score (0 to 1) for each feature
# Higher = that feature was used more often and at higher-level splits
# All values sum to 1.0

rf_model       = pipe_rf.named_steps['rf']
feature_names  = iris.feature_names
importances    = rf_model.feature_importances_
std_of_trees   = np.std([tree.feature_importances_ for tree in rf_model.estimators_], axis=0)

print("═══ FEATURE IMPORTANCE SCORES ═══")
print()
print(f"{'Rank':<6} {'Feature':<28} {'Importance':>12} {'Visualised'}")
print("─" * 65)

# Sort features by importance
sorted_idx = np.argsort(importances)[::-1]  # descending order
for rank, idx in enumerate(sorted_idx, 1):
    bar = "█" * int(importances[idx] * 50)
    print(f"{rank:<6} {feature_names[idx]:<28} {importances[idx]:>12.4f}  {bar}")

print()
print("All importances sum to:", importances.sum().round(4), "(should be 1.0)")
print()
print("INTERPRETATION:")
print("  Features with HIGH importance are the ones the forest relies on most.")
print("  Features with near-ZERO importance could potentially be removed.")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 26 — FEATURE IMPORTANCE VISUALIZATION
# ═══════════════════════════════════════════════════════════════════

fig, axes = plt.subplots(1, 2, figsize=(15, 6))
fig.suptitle('Random Forest — Feature Importance Analysis', fontsize=13, fontweight='bold')

sorted_idx = np.argsort(importances)   # ascending (for horizontal bar chart)

# ── LEFT: Horizontal bar chart ──────────────────────────────────
colors_fi = ['#E74C3C' if i in sorted_idx[-2:] else '#3498DB' for i in range(4)]
colors_fi_sorted = [colors_fi[i] for i in sorted_idx]

axes[0].barh(
    [feature_names[i] for i in sorted_idx],  # feature names on y-axis
    importances[sorted_idx],                  # importance values
    color=colors_fi_sorted,
    alpha=0.85,
    edgecolor='white',
    linewidth=0.5
)
# Add value labels
for i, (imp, idx) in enumerate(zip(importances[sorted_idx], sorted_idx)):
    axes[0].text(imp + 0.005, i, f'{imp:.4f}', va='center', fontsize=10)

axes[0].set_xlabel('Feature Importance Score', fontsize=11)
axes[0].set_title('Feature Importance (Red = top 2 most important features)', fontsize=11, fontweight='bold')
axes[0].grid(axis='x', alpha=0.3)
axes[0].set_xlim([0, max(importances) * 1.3])

# ── RIGHT: Number of trees vs accuracy ──────────────────────────
n_trees_range = [1, 5, 10, 20, 50, 100, 150, 200]
tree_test_accs  = []
tree_train_accs = []

for n in n_trees_range:
    rf_n = Pipeline([
        ('s', StandardScaler()),
        ('m', RandomForestClassifier(n_estimators=n, random_state=42, n_jobs=-1))
    ])
    rf_n.fit(X_train, y_train)
    tree_train_accs.append(accuracy_score(y_train, rf_n.predict(X_train)))
    tree_test_accs.append(accuracy_score(y_test,   rf_n.predict(X_test)))

axes[1].plot(n_trees_range, tree_train_accs, 'o-', color='#3498DB', lw=2, ms=7, label='Training')
axes[1].plot(n_trees_range, tree_test_accs,  's-', color='#E74C3C', lw=2, ms=7, label='Test')
axes[1].set_xlabel('Number of Trees (n_estimators)', fontsize=11)
axes[1].set_ylabel('Accuracy', fontsize=11)
axes[1].set_title('More Trees = Better (up to a point) Accuracy vs Number of Trees',
                  fontsize=11, fontweight='bold')
axes[1].legend(fontsize=10); axes[1].grid(alpha=0.3)
axes[1].set_ylim([0.90, 1.01])

plt.tight_layout()
plt.savefig('tmp/random_forest.png', dpi=100, bbox_inches='tight')
plt.show()

print("KEY INSIGHT from the right chart:")
print("  Adding more trees initially improves accuracy — ensemble effect.")
print("  After ~50 trees, accuracy plateaus. More trees = more computation.")
print("  100–200 trees is usually the sweet spot (good accuracy, reasonable speed).")
````

## 📈 Section 12 · Regression — Predicting Numbers (Diabetes Dataset)

### Classification vs Regression

So far we have been predicting **categories** (which species? malignant or benign?). This is **classification**.

Now we switch to **regression** — predicting a **continuous number**.

| | Classification | Regression |
|--|----------------|-----------|
| Output | A category | A number |
| Example | "This email is spam" | "This house costs ₦45 million" |
| Example | "This tumour is malignant" | "This patient's blood sugar is 7.2" |
| Metric | Accuracy, F1 | MAE, RMSE, R² |

### The Diabetes Dataset

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 29 — REGRESSION MODELS: TRAIN AND COMPARE
# ═══════════════════════════════════════════════════════════════════

from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

# Prepare data
X_d = diabetes.data
y_d = diabetes.target

# Split
X_d_train, X_d_test, y_d_train, y_d_test = train_test_split(
    X_d, y_d, test_size=0.2, random_state=42
)

print(f"Training samples: {len(X_d_train)}")
print(f"Test samples    : {len(X_d_test)}")
print()

# ── Define regression models ────────────────────────────────────
regression_models = {
    'Linear Regression'     : LinearRegression(),
    'Ridge (alpha=1.0)'     : Ridge(alpha=1.0),
    'Lasso (alpha=0.1)'     : Lasso(alpha=0.1, max_iter=5000),
    'Decision Tree (d=4)'   : DecisionTreeRegressor(max_depth=4, random_state=42),
    'Random Forest (100)'   : RandomForestRegressor(n_estimators=100, random_state=42),
    'Gradient Boosting'     : GradientBoostingRegressor(n_estimators=100, random_state=42),
}

# ── REGRESSION METRIC EXPLANATIONS ─────────────────────────────
print("METRICS WE WILL USE:")
print()
print("  MAE  (Mean Absolute Error):")
print("       Average of |actual - predicted|")
print("       If MAE=40, our prediction is on average 40 points off.")
print("       Same units as the target. Easy to interpret.")
print()
print("  RMSE (Root Mean Squared Error):")
print("       √ mean( (actual - predicted)² )")
print("       Penalises LARGE errors more heavily than MAE.")
print("       Also in target units. Prefer MAE for explainability.")
print()
print("  R²   (R-squared / Coefficient of Determination):")
print("       How much of the VARIANCE in the target does the model explain?")
print("       1.0 = perfect predictions. 0.0 = as good as predicting the mean.")
print("       Negative = model is WORSE than the mean (very bad!)")
print()
print("─" * 65)
print(f"{'Model':<26} {'MAE':>8} {'RMSE':>8} {'R²':>8}")
print("─" * 65)

reg_results = {}

for name, model in regression_models.items():
    # Fit (note: diabetes data is already scaled, no StandardScaler needed)
    model.fit(X_d_train, y_d_train)
    y_pred_reg = model.predict(X_d_test)

    mae  = mean_absolute_error(y_d_test, y_pred_reg)
    rmse = np.sqrt(mean_squared_error(y_d_test, y_pred_reg))
    r2   = r2_score(y_d_test, y_pred_reg)

    reg_results[name] = {'mae': mae, 'rmse': rmse, 'r2': r2, 'model': model, 'pred': y_pred_reg}

    # Visual indicator
    indicator = "⭐" if r2 == max(v['r2'] for v in reg_results.values()) else "  "
    print(f"{indicator} {name:<24} {mae:>8.2f} {rmse:>8.2f} {r2:>8.4f}")

print("─" * 65)
best_reg = max(reg_results, key=lambda k: reg_results[k]['r2'])
print()
print(f"Best model: {best_reg}")
print(f"  R² = {reg_results[best_reg]['r2']:.4f} (explains {reg_results[best_reg]['r2']*100:.1f}% of variance)")
print(f"  MAE = {reg_results[best_reg]['mae']:.2f} (average prediction error)")
````

## 🎛️ Section 14 · Hyperparameter Tuning — Finding the Best Settings

### What Are Hyperparameters?

Every ML algorithm has **settings you choose before training** — these are called **hyperparameters**:

- KNN: `n_neighbors` (K)
- Decision Tree: `max_depth`, `min_samples_split`
- Random Forest: `n_estimators`, `max_depth`, `max_features`
- SVM: `C`, `gamma`, `kernel`
- Logistic Regression: `C`, `solver`

Choosing good hyperparameters can dramatically improve performance. But how do you know which values are best?

### Method 1: GridSearchCV — Try Every Combination

Define a grid of values. Try every possible combination. Pick the winner.

```
param_grid = {
    'n_estimators': [50, 100, 200],    → 3 options
    'max_depth':    [None, 5, 10],     → 3 options
}
Grid size = 3 × 3 = 9 combinations
Each evaluated with 5-fold CV = 9 × 5 = 45 model fits
```

**Pro:** Guaranteed to find the best combination in the grid
**Con:** Gets slow quickly with many parameters

### Method 2: RandomizedSearchCV — Sample the Space

Instead of trying everything, randomly sample from distributions.

**Pro:** Much faster, often finds nearly-as-good solutions
**Con:** Might miss the absolute best combination

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 35 — GRIDSEARCHCV: EXHAUSTIVE HYPERPARAMETER SEARCH
# ═══════════════════════════════════════════════════════════════════

from sklearn.model_selection import GridSearchCV

# We will tune a Random Forest on the Breast Cancer dataset
# Using the cancer train/test split from Section 13

# ── Define the parameter grid ────────────────────────────────────
# Keys: must use format 'step_name__param_name' for pipeline parameters
# (double underscore connects the pipeline step name to the parameter name)
param_grid = {
    'model__n_estimators'     : [50, 100, 200],         # number of trees
    'model__max_depth'        : [None, 5, 10, 20],       # max tree depth
    'model__min_samples_split': [2, 5, 10],              # min samples to split a node
}

# Total combinations: 3 × 4 × 3 = 36
# Each with 5-fold CV: 36 × 5 = 180 model fits
print(f"Grid combinations: {3 * 4 * 3} = {3}×{4}×{3}")
print(f"Total fits with 5-fold CV: {3 * 4 * 3 * 5}")
print()

# Create the base pipeline
pipe_for_grid = Pipeline([
    ('scaler', StandardScaler()),
    ('model',  RandomForestClassifier(random_state=42, n_jobs=-1))
])

# GridSearchCV wraps our pipeline and tries every combination
grid_search = GridSearchCV(
    estimator  = pipe_for_grid,          # the pipeline to tune
    param_grid = param_grid,             # hyperparameter grid
    cv         = StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
    scoring    = 'f1_weighted',          # optimise for F1 (better than accuracy for imbalanced data)
    n_jobs     = -1,                     # use all CPU cores
    verbose    = 0,                      # suppress progress messages
    return_train_score = True            # also record training scores
)

print("Running Grid Search... (this tries 180 model fits, may take ~30 seconds)")
grid_search.fit(X_c_train, y_c_train)

print()
print("═══ GRID SEARCH RESULTS ═══")
print()
print("Best hyperparameters found:")
for param, value in grid_search.best_params_.items():
    # Clean up the parameter name for display
    clean_name = param.replace('model__', '')
    print(f"  {clean_name:<24}: {value}")

print()
print(f"Best cross-validation F1 score: {grid_search.best_score_:.4f}")
print()

# Evaluate the best model on the test set
best_model_gs = grid_search.best_estimator_
y_gs_pred = best_model_gs.predict(X_c_test)
print(f"Test accuracy (best model)  : {accuracy_score(y_c_test, y_gs_pred):.4f}")
print(f"Test F1 weighted (best)     : {grid_search.score(X_c_test, y_c_test):.4f}")
print()
print(classification_report(y_c_test, y_gs_pred, target_names=cancer.target_names))
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 36 — RANDOMIZEDSEARCHCV: FASTER ALTERNATIVE
# ═══════════════════════════════════════════════════════════════════

from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint, uniform

# Instead of fixed lists, we define DISTRIBUTIONS to sample from
# randint(low, high): random integer between low and high-1
# uniform(low, scale): random float between low and low+scale

param_distributions = {
    'model__n_estimators'    : randint(50, 500),         # any integer 50-499
    'model__max_depth'       : [None, 3, 5, 7, 10, 15, 20],
    'model__min_samples_split': randint(2, 20),           # any integer 2-19
    'model__min_samples_leaf' : randint(1, 10),
    'model__max_features'    : ['sqrt', 'log2', None],
}

# RandomizedSearchCV samples n_iter random combinations
random_search = RandomizedSearchCV(
    estimator           = pipe_for_grid,
    param_distributions = param_distributions,
    n_iter              = 40,       # only try 40 random combinations (vs 36 exhaustive above)
    cv                  = StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
    scoring             = 'f1_weighted',
    n_jobs              = -1,
    random_state        = 42,       # reproducible random sampling
    verbose             = 0,
    return_train_score  = True
)

print("Running Randomized Search... (40 combinations × 5-fold = 200 fits)")
random_search.fit(X_c_train, y_c_train)

print()
print("═══ RANDOMIZED SEARCH RESULTS ═══")
print()
print("Best hyperparameters found:")
for param, value in random_search.best_params_.items():
    clean_name = param.replace('model__', '')
    print(f"  {clean_name:<24}: {value}")

print()
print(f"Best CV F1 score     : {random_search.best_score_:.4f}")

y_rs_pred = random_search.best_estimator_.predict(X_c_test)
print(f"Test accuracy        : {accuracy_score(y_c_test, y_rs_pred):.4f}")

print()
print("═══ COMPARISON: GRID SEARCH vs RANDOMIZED SEARCH ═══")
print()
print(f"  Grid Search      — {3*4*3} combinations, Best CV F1: {grid_search.best_score_:.4f}, Test Acc: {accuracy_score(y_c_test, y_gs_pred):.4f}")
print(f"  Randomized Search — 40 combinations, Best CV F1: {random_search.best_score_:.4f}, Test Acc: {accuracy_score(y_c_test, y_rs_pred):.4f}")
print()
print("CONCLUSION:")
print("  RandomizedSearch often finds equally good results with much less computation.")
print("  GridSearch guarantees you try every combination (useful when grid is small).")
print("  For large parameter spaces: ALWAYS use RandomizedSearch.")
````

## 🔢 Section 15 · Bonus — Handwritten Digit Recognition

### A Bigger Real-World Problem

The **Digits dataset** contains **1,797 grayscale images** of handwritten digits (0-9) collected by researchers. Each image is 8×8 pixels = **64 features** per image.

This is similar to the famous MNIST dataset used to benchmark computer vision algorithms.

**The Challenge:** Given a 64-pixel image, which digit is it (0-9)?

This is a **10-class** classification problem — much harder than Iris's 3 classes!

```
Example digit "3":
  [ 0  0  5 13  9  1  0  0]   ← 8 pixel values in row 1
  [ 0  0 13 15 10 15  5  0]   ← 8 pixel values in row 2
  [ 0  3 15  2  0 11  8  0]   ← etc.
  [ 0  4 12  0  0  8  8  0]
  [ 0  5  8  0  0  9  8  0]
  [ 0  4 11  0  1 12  7  0]
  [ 0  2 14  5 10 12  0  0]
  [ 0  0  6 13 10  0  0  0]
  Values 0-16: 0=white, 16=black
```

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 37 — DIGIT RECOGNITION: LOADING AND VISUALISING
# ═══════════════════════════════════════════════════════════════════

from sklearn.datasets import load_digits

# Load the real handwritten digits dataset
digits = load_digits()

print("═══ DIGITS DATASET OVERVIEW ═══")
print()
print(f"Number of images    : {digits.data.shape[0]}")
print(f"Features per image  : {digits.data.shape[1]} (8×8 = 64 pixels)")
print(f"Number of classes   : {len(digits.target_names)} (digits 0-9)")
print(f"Image shape         : {digits.images.shape}")
print()

# Each image is 8x8 pixels with values 0 (white) to 16 (black)
print("Example: first image data (digit:", digits.target[0], ")")
print(digits.images[0].astype(int))

# ── Visualise some real handwritten digit images ────────────────
fig, axes = plt.subplots(3, 10, figsize=(16, 6))
fig.suptitle('Real Handwritten Digits from the Dataset '
             '(Each image is 8×8 pixels · Grey shading = pixel intensity)',
             fontsize=12, fontweight='bold')

for digit_class in range(10):
    # Find 3 examples of this digit class
    indices = np.where(digits.target == digit_class)[0][:3]
    for row, idx in enumerate(indices):
        axes[row, digit_class].imshow(
            digits.images[idx],    # the 8x8 pixel grid
            cmap='gray_r',         # grey reversed: black digit on white background
            interpolation='none'   # no smoothing — show raw pixels
        )
        axes[row, digit_class].axis('off')   # hide axis ticks
        if row == 0:
            axes[row, digit_class].set_title(f'Digit {digit_class}', fontsize=10)

plt.tight_layout()
plt.savefig('tmp/digits_gallery.png', dpi=100, bbox_inches='tight')
plt.show()

print("Each column shows 3 DIFFERENT people's handwriting of the same digit.")
print("The algorithm must learn to recognise digit regardless of handwriting style!")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 38 — DIGIT RECOGNITION: BUILDING AND EVALUATING MODELS
# ═══════════════════════════════════════════════════════════════════

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay

# Prepare data
X_dig = digits.data     # shape: (1797, 64)
y_dig = digits.target   # shape: (1797,) — values 0-9

# Split
X_dig_train, X_dig_test, y_dig_train, y_dig_test = train_test_split(
    X_dig, y_dig, test_size=0.2, random_state=42, stratify=y_dig
)

print(f"Training: {len(X_dig_train)} images | Test: {len(X_dig_test)} images")
print()

# Compare models on digit recognition
digit_models = {
    'KNN (k=5)'          : KNeighborsClassifier(n_neighbors=5),
    'Logistic Regression': LogisticRegression(max_iter=2000, random_state=42),
    'Random Forest (100)': RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    'SVM (RBF)'          : SVC(kernel='rbf', C=10, gamma='scale', random_state=42),
}

print(f"{'Model':<26} {'Train Acc':>12} {'Test Acc':>12}")
print("─" * 54)

digit_results = {}
for name, model in digit_models.items():
    pipe_dig = Pipeline([
        ('scaler', StandardScaler()),
        ('model',  model)
    ])
    pipe_dig.fit(X_dig_train, y_dig_train)

    train_acc = accuracy_score(y_dig_train, pipe_dig.predict(X_dig_train))
    test_acc  = accuracy_score(y_dig_test,  pipe_dig.predict(X_dig_test))
    digit_results[name] = {'pipe': pipe_dig, 'test_acc': test_acc}

    print(f"  {name:<24} {train_acc:>12.4f} {test_acc:>12.4f}")

# Use best model for further analysis
best_digit_name = max(digit_results, key=lambda k: digit_results[k]['test_acc'])
best_digit_pipe = digit_results[best_digit_name]['pipe']
y_dig_pred = best_digit_pipe.predict(X_dig_test)

print()
print(f"Best model: {best_digit_name}")
print(f"Test accuracy: {digit_results[best_digit_name]['test_acc']:.4f}")
print()

# Confusion matrix for digit classification (10x10!)
fig, ax = plt.subplots(figsize=(9, 8))
cm_dig = confusion_matrix(y_dig_test, y_dig_pred)
ConfusionMatrixDisplay(cm_dig, display_labels=digits.target_names).plot(
    ax=ax, cmap='Blues', colorbar=True
)
ax.set_title(f'Digit Recognition — Confusion Matrix {best_digit_name} | Test Accuracy = {digit_results[best_digit_name]["test_acc"]:.4f}',
             fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig('tmp/digits_cm.png', dpi=100, bbox_inches='tight')
plt.show()

print("READ THE CONFUSION MATRIX:")
print("  Diagonal = correct predictions (we want HIGH numbers here)")
print("  Off-diagonal = confusions (e.g., mistaking 3 for 8)")
print()
print("  Common confusions in handwritten digits:")
print("  3 ↔ 8  (similar curved strokes)")
print("  4 ↔ 9  (similar vertical line structure)")
print("  1 ↔ 7  (similar upstroke)")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 39 — DIGIT RECOGNITION: VISUALISING PREDICTIONS
# ═══════════════════════════════════════════════════════════════════

# Show what the model predicts for individual test images
# This helps us understand WHERE and WHY it makes mistakes

fig, axes = plt.subplots(4, 8, figsize=(16, 9))
fig.suptitle(f'Digit Recognition Predictions — {best_digit_name} '
             '(Green border = Correct · Red border = Wrong)',
             fontsize=12, fontweight='bold')

y_dig_pred_all = best_digit_pipe.predict(X_dig_test)

# Show 32 test images
for i, ax in enumerate(axes.flatten()):
    # Get the image (reshape from 64 values to 8x8 grid)
    img = X_dig_test[i].reshape(8, 8)

    ax.imshow(img, cmap='gray_r', interpolation='none')
    ax.axis('off')

    predicted = y_dig_pred_all[i]
    actual    = y_dig_test[i]
    correct   = (predicted == actual)

    # Green title if correct, red if wrong
    color  = '#27AE60' if correct else '#E74C3C'
    label  = f"P:{predicted} A:{actual}" if not correct else f"✓ {actual}"
    weight = 'bold'

    ax.set_title(label, fontsize=8, color=color, fontweight=weight)

    # Add coloured border
    for spine in ax.spines.values():
        spine.set_edgecolor(color)
        spine.set_linewidth(2.5)
        spine.set_visible(True)

plt.tight_layout()
plt.savefig('tmp/digit_predictions.png', dpi=100, bbox_inches='tight')
plt.show()

n_correct = np.sum(y_dig_pred_all == y_dig_test)
n_total   = len(y_dig_test)
n_wrong   = n_total - n_correct
print(f"In these {n_total} test images:")
print(f"  ✅ Correct : {n_correct}")
print(f"  ❌ Wrong   : {n_wrong}")
print()
print("Look for the RED-bordered images — those are the model's mistakes.")
print("Notice which digits are being confused with which other digits.")
print("These confusions often make visual sense!")
````

## 💾 Section 16 · Saving and Loading Trained Models

### Why We Save Models

Training a machine learning model takes time and computational resources. Once you have a well-performing model, you want to:

1. **Save it** once after training
2. **Load it** every time you need to make predictions (without retraining!)

### The joblib Library

`joblib` is the recommended way to save sklearn models to disk. It handles large NumPy arrays (like the 200 decision trees in a Random Forest) efficiently.

### Production Workflow

```
DEVELOPMENT (once):                  PRODUCTION (every time):
  Train model                          Load saved model
  Tune hyperparameters      ──────→    Receive new data
  Evaluate performance                 Make prediction in milliseconds
  Save model + metadata                Return result to user
```

### What to Save
1. **The model file** (`.joblib`) — the trained model object
2. **Metadata file** (`.json`) — version, accuracy, training info

This way, if something goes wrong in production, you know exactly which version of the model is running.

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 40 — SAVING A TRAINED MODEL
# ═══════════════════════════════════════════════════════════════════

import joblib          # handles saving/loading Python objects to disk
import json            # handles saving/loading JSON files
import os              # for file operations
from datetime import datetime   # for timestamps

# ── Use the best breast cancer model from Section 13 ────────────
# (Already trained: best_model_gs from GridSearchCV)

# STEP 1: Make sure we have a final trained model
final_model = best_model_gs   # the GridSearch winner (pipeline: scaler + RandomForest)

# Quick verification
final_train_acc = accuracy_score(y_c_train, final_model.predict(X_c_train))
final_test_acc  = accuracy_score(y_c_test,  final_model.predict(X_c_test))

print("═══ FINAL MODEL VERIFICATION ═══")
print()
print(f"  Training accuracy : {final_train_acc:.4f}")
print(f"  Test accuracy     : {final_test_acc:.4f}")
print()

# STEP 2: Save the model file using joblib
model_filepath = 'tmp/breast_cancer_model_v1.joblib'
joblib.dump(final_model, model_filepath)

file_size_kb = os.path.getsize(model_filepath) / 1024
print(f"Model saved to  : {model_filepath}")
print(f"File size       : {file_size_kb:.1f} KB")
print()

# STEP 3: Save metadata — ALWAYS do this!
# This file explains what the model is and how it was trained
metadata = {
    'model_name'         : 'Breast Cancer Random Forest Classifier',
    'version'            : '1.0',
    'algorithm'          : 'RandomForestClassifier',
    'training_date'      : datetime.now().strftime('%Y-%m-%d %H:%M:%S'),
    'sklearn_version'    : sklearn.__version__,
    'n_training_samples' : len(X_c_train),
    'n_test_samples'     : len(X_c_test),
    'n_features'         : X_c.shape[1],
    'feature_names'      : list(cancer.feature_names),
    'target_classes'     : list(cancer.target_names),
    'hyperparameters'    : {
        k.replace('model__', ''): v
        for k, v in grid_search.best_params_.items()
    },
    'performance': {
        'training_accuracy': round(final_train_acc, 4),
        'test_accuracy'    : round(final_test_acc, 4),
    }
}

metadata_filepath = '/tmp/breast_cancer_model_v1_metadata.json'
with open(metadata_filepath, 'w') as f:
    json.dump(metadata, f, indent=2)   # indent=2 makes JSON human-readable

print(f"Metadata saved to: {metadata_filepath}")
print()
print("Metadata contents:")
print(json.dumps(metadata, indent=2))
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 41 — LOADING A SAVED MODEL AND MAKING PREDICTIONS
# ═══════════════════════════════════════════════════════════════════

# This simulates what happens in PRODUCTION
# The model was saved once, now we load it and use it on new data

print("═══ LOADING AND USING THE SAVED MODEL ═══")
print()

# STEP 1: Load the model from disk
# This works in ANY Python session — the model does not need to be retrained
loaded_model = joblib.load(model_filepath)
print(f"✅ Model loaded from: {model_filepath}")
print()

# STEP 2: Verify it gives the same predictions as the original
original_predictions = final_model.predict(X_c_test)
loaded_predictions   = loaded_model.predict(X_c_test)

predictions_match = np.array_equal(original_predictions, loaded_predictions)
print(f"Predictions match original: {predictions_match} ✅")
print()

# STEP 3: Load and display metadata
with open(metadata_filepath, 'r') as f:
    loaded_metadata = json.load(f)

print(f"Model name    : {loaded_metadata['model_name']}")
print(f"Version       : {loaded_metadata['version']}")
print(f"Training date : {loaded_metadata['training_date']}")
print(f"Test accuracy : {loaded_metadata['performance']['test_accuracy']}")
print()

# ── Simulate real usage: predict on a new patient ─────────────
print("═══ SIMULATING REAL USAGE: NEW PATIENT ═══")
print()

# Use the first 3 test patients as "new patients"
new_patients = X_c_test[:3]
true_labels  = y_c_test[:3]

print("Processing 3 new patients...")
print()

predictions   = loaded_model.predict(new_patients)
probabilities = loaded_model.predict_proba(new_patients)

for i, (pred, probs, actual) in enumerate(zip(predictions, probabilities, true_labels)):
    diagnosis_pred   = cancer.target_names[pred]
    diagnosis_actual = cancer.target_names[actual]
    confidence       = probs[pred] * 100
    correct          = "✅" if pred == actual else "❌"

    print(f"  Patient {i+1}:")
    print(f"    Prediction  : {diagnosis_pred}  (confidence: {confidence:.1f}%)")
    print(f"    Probability benign   : {probs[1]*100:.1f}%")
    print(f"    Probability malignant: {probs[0]*100:.1f}%")
    print(f"    Actual      : {diagnosis_actual}  {correct}")
    print()

print("This is how a deployed model works:")
print("  1. Hospital system receives new patient measurements")
print("  2. Data is fed to loaded_model.predict(new_measurements)")
print("  3. Prediction is returned in milliseconds")
print("  4. Doctor sees the AI's recommendation alongside their own assessment")
````

````python
# ═══════════════════════════════════════════════════════════════════
#  CELL 43 — WINE PROJECT: PCA VISUALISATION
#  (How to visualise HIGH-DIMENSIONAL data!)
# ═══════════════════════════════════════════════════════════════════

# The wine data has 13 features — we cannot plot 13 dimensions!
# PCA (Principal Component Analysis) solves this:
# It finds the 2 "directions" that capture the MOST VARIATION in the data
# and projects all 13D points onto these 2 directions.
# Result: we can plot the data in 2D while preserving the most important patterns.

# This is NOT what we use for the model — just for VISUALISATION.

from sklearn.decomposition import PCA

# Always scale before PCA — features with large numbers would dominate!
scaler_pca = StandardScaler()
X_wine_scaled = scaler_pca.fit_transform(wine.data)

# PCA: reduce 13 dimensions to 2
pca = PCA(n_components=2, random_state=42)
X_wine_2d = pca.fit_transform(X_wine_scaled)

# How much of the original information is preserved?
print(f"Variance explained by PC1: {pca.explained_variance_ratio_[0]*100:.1f}%")
print(f"Variance explained by PC2: {pca.explained_variance_ratio_[1]*100:.1f}%")
print(f"Total variance retained  : {sum(pca.explained_variance_ratio_)*100:.1f}%")
print()
print(f"We compressed 13 dimensions to 2 while keeping {sum(pca.explained_variance_ratio_)*100:.1f}% of information!")
print()

# Plot
fig, axes = plt.subplots(1, 2, figsize=(15, 6))
fig.suptitle('Wine Dataset — Visualising 13 Features in 2D using PCA',
             fontsize=13, fontweight='bold')

colors_wine  = ['#E74C3C', '#3498DB', '#2ECC71']
markers_wine = ['o', 's', '^']

# ── Left: PCA plot ──────────────────────────────────────────────
for i, (name, col, mkr) in enumerate(zip(wine.target_names, colors_wine, markers_wine)):
    mask = wine.target == i
    axes[0].scatter(
        X_wine_2d[mask, 0], X_wine_2d[mask, 1],
        c=col, marker=mkr, s=60, alpha=0.8,
        label=f'Cultivar {name}', edgecolors='white', linewidths=0.5
    )

axes[0].set_xlabel(f'Principal Component 1 ({pca.explained_variance_ratio_[0]*100:.1f}% of variance)', fontsize=10)
axes[0].set_ylabel(f'Principal Component 2 ({pca.explained_variance_ratio_[1]*100:.1f}% of variance)', fontsize=10)
axes[0].set_title('PCA: All 13 Features → 2D '
                  'Clusters = different wine cultivars', fontsize=11, fontweight='bold')
axes[0].legend(fontsize=10); axes[0].grid(alpha=0.3)

# ── Right: Feature importance for PCA ───────────────────────────
# Which original features contribute most to PC1?
pc1_loadings = pd.Series(np.abs(pca.components_[0]), index=wine.feature_names)
pc1_loadings_sorted = pc1_loadings.sort_values(ascending=True)

axes[1].barh(pc1_loadings_sorted.index, pc1_loadings_sorted.values,
             color='#9B59B6', alpha=0.8)
axes[1].set_xlabel('Absolute Loading (contribution to PC1)', fontsize=10)
axes[1].set_title('Feature Contributions to PC1 '
                  '(Higher = more important for separating wine types)', fontsize=11, fontweight='bold')
axes[1].grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.savefig('tmp/wine_pca.png', dpi=100, bbox_inches='tight')
plt.show()

print("The PCA plot shows:")
print("  Three clear clusters = wine cultivars are chemically distinct")
print("  Good separation = machine learning should classify these well")
````

### 📊 Metrics Quick Reference

**Classification Metrics:**
```
Accuracy  = (Correct Predictions) / (Total Predictions)
           → Use when classes are balanced

Precision = TP / (TP + FP)
           → "When I say positive, how often am I right?"
           → Use when false alarms are costly

Recall    = TP / (TP + FN)
           → "Of all real positives, how many did I catch?"
           → Use when MISSING a case is costly (cancer, fraud)

F1-Score  = 2 × (Precision × Recall) / (Precision + Recall)
           → Balance of precision and recall
           → Use with imbalanced classes

ROC-AUC   → Overall ranking quality, independent of threshold
           → 1.0 = perfect, 0.5 = random, higher = better
```

**Regression Metrics:**
```
MAE  (Mean Absolute Error) = mean(|actual - predicted|)
     → Average error in same units as target. Easy to explain.

RMSE (Root Mean Squared Error) = sqrt(mean((actual-predicted)²))
     → Penalises large errors more than MAE. Same units as target.

R²   (R-squared) = 1 - SS_residuals / SS_total
     → 1.0 = perfect, 0.0 = as good as predicting mean, negative = terrible
```

### ⚠️ The 7 Golden Rules of Machine Learning

1. **NEVER evaluate on training data** → Use a held-out test set
2. **Split BEFORE preprocessing** → Fit scaler on training only
3. **Use Pipelines** → They make #2 automatic and safe
